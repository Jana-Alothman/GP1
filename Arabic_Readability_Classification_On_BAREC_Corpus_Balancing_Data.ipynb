- ID: Unique sentence identifier.
- Sentence: The sentence text.

- Word: Simply tokenized and dediacritized sentences.
- Lex: Each word is replaced by its predicited lemma (dediacritized).
- D3Tok: We tokenize words into their base and clitics forms.
- D3Lex: We replace the base forms in D3Tok with the predicited lemmas.
- Readability_Level: The readability level in 19-levels scheme, ranging from 1-alif to 19-qaf.

- Readability_Level_5: The readability level in 5-levels scheme, ranging from 1 to 5.
- Readability_Level_3: The readability level in 3-levels scheme, ranging from 1 to 3.
- Annotator: The annotator ID (A1-A5 or IAA).
- Document: Source document file name.
- Source: Document source.
- Book: Book name.
- Author: Author name.
- Domain: Domain (Arts & Humanities, STEM or Social Sciences).
- Text_Class: Readership group (Foundational, Advanced or Specialized).

---

The dataset supports multi-class readability classification in the following formats:

- 19 levels (default)
- 7 levels
- 5 levels
- 3 levels

## Step 4: Determine Augmentation Targets

Based on the training set distribution, the dataset shows substantial class imbalance, particularly for `Readability_Level_19`.

Instead of forcing all classes to have the same number of samples, a moderate target is considered to reduce the imbalance while avoiding excessive synthetic data generation.

The target will be applied only to minority classes, while classes already above the target will remain unchanged.

```python
# ============================================================
# STEP 4: DETERMINE AUGMENTATION TARGETS
# ============================================================

import numpy as np
import pandas as pd

def determine_augmentation_targets(train_df, label_column, strategy='median'):
    """
    Determine target number of samples per class for augmentation.

    Parameters:
    - train_df: Training DataFrame
    - label_column: Name of the label column
    - strategy: Target strategy ('median', 'mean', 'quantile', or custom integer)

    Returns:
    - targets: Dictionary with target count per class
    - current_counts: Series with current count per class
    - needed_augmentation: Dictionary with needed augmentation per class
    """

    # Get current distribution
    current_counts = train_df[label_column].value_counts().sort_index()

    if strategy == 'median':
        target = int(np.median(current_counts))
        print(f"Using MEDIAN as target: {target} samples per class")

    elif strategy == 'mean':
        target = int(np.mean(current_counts))
        print(f"Using MEAN as target: {target} samples per class")

    elif strategy == 'quantile':
        target = int(np.quantile(current_counts, 0.75))
        print(f"Using 75th PERCENTILE as target: {target} samples per class")

    elif isinstance(strategy, int):
        target = strategy
        print(f"Using CUSTOM target: {target} samples per class")

    else:
        raise ValueError("Strategy must be 'median', 'mean', 'quantile', or an integer")

    # Determine targets for each class
    targets = {}
    needed_augmentation = {}

    for class_label, current_count in current_counts.items():
        if current_count < target:
            targets[class_label] = target
            needed_augmentation[class_label] = target - current_count
        else:
            targets[class_label] = current_count
            needed_augmentation[class_label] = 0

    return targets, current_counts, needed_augmentation


# ============================================================
# APPLY TO READABILITY_LEVEL_19
# ============================================================

print("=" * 60)
print("READABILITY LEVEL 19 - Augmentation Targets")
print("=" * 60)

targets_19, current_counts_19, needed_aug_19 = determine_augmentation_targets(
    train_19,
    "Readability_Level_19",
    strategy='median'
)

# Create results DataFrame
results_19 = pd.DataFrame({
    'Current_Count': current_counts_19.values,
    'Target_Count': [targets_19[idx] for idx in current_counts_19.index],
    'Needed_Augmentation': [needed_aug_19[idx] for idx in current_counts_19.index],
    'Augmentation_Needed': ['Yes' if needed_aug_19[idx] > 0 else 'No' for idx in current_counts_19.index]
}, index=current_counts_19.index)

print("\nDistribution Summary:")
display(results_19)

# Statistics
total_aug_needed_19 = sum(needed_aug_19.values())
classes_to_augment_19 = sum(1 for v in needed_aug_19.values() if v > 0)

print(f"\nStatistics:")
print(f"   Classes that need augmentation: {classes_to_augment_19} out of {len(current_counts_19)}")
print(f"   Total new samples to generate: {total_aug_needed_19:,}")
print(f"   Training set after augmentation: {len(train_19) + total_aug_needed_19:,}")


# ============================================================
# APPLY TO READABILITY_LEVEL_5
# ============================================================

print("\n" + "=" * 60)
print("READABILITY LEVEL 5 - Augmentation Targets")
print("=" * 60)

targets_5, current_counts_5, needed_aug_5 = determine_augmentation_targets(
    train_5,
    "Readability_Level_5",
    strategy='median'
)

# Create results DataFrame
results_5 = pd.DataFrame({
    'Current_Count': current_counts_5.values,
    'Target_Count': [targets_5[idx] for idx in current_counts_5.index],
    'Needed_Augmentation': [needed_aug_5[idx] for idx in current_counts_5.index],
    'Augmentation_Needed': ['Yes' if needed_aug_5[idx] > 0 else 'No' for idx in current_counts_5.index]
}, index=current_counts_5.index)

print("\nDistribution Summary:")
display(results_5)

# Statistics
total_aug_needed_5 = sum(needed_aug_5.values())
classes_to_augment_5 = sum(1 for v in needed_aug_5.values() if v > 0)

print(f"\nStatistics:")
print(f"   Classes that need augmentation: {classes_to_augment_5} out of {len(current_counts_5)}")
print(f"   Total new samples to generate: {total_aug_needed_5:,}")
print(f"   Training set after augmentation: {len(train_5) + total_aug_needed_5:,}")


# ============================================================
# COMPARE DIFFERENT TARGET STRATEGIES
# ============================================================

def analyze_target_strategy(train_df, label_column):
    """Compare different target strategies for augmentation"""

    current_counts = train_df[label_column].value_counts().sort_index()

    strategies = {
        'Median': int(np.median(current_counts)),
        'Mean': int(np.mean(current_counts))
    }

    print("\n" + "=" * 60)
    print("COMPARING DIFFERENT TARGET STRATEGIES")
    print("=" * 60)

    comparison_data = {
        'Strategy': [],
        'Target': [],
        'Classes_to_Augment': [],
        'Total_New_Samples': []
    }

    for strategy_name, target_value in strategies.items():
        comparison_data['Strategy'].append(strategy_name)
        comparison_data['Target'].append(target_value)
        comparison_data['Classes_to_Augment'].append(
            sum(1 for c in current_counts if c < target_value)
        )
        comparison_data['Total_New_Samples'].append(
            sum(max(0, target_value - c) for c in current_counts)
        )

    comparison_df = pd.DataFrame(comparison_data)
    display(comparison_df)


# Analyze for Readability_Level_19
analyze_target_strategy(train_19, "Readability_Level_19")

# Analyze for Readability_Level_5
analyze_target_strategy(train_5, "Readability_Level_5")
```

# Step 5.1: Arabic Contextual Augmentation (AraBERT)

## Overview
Uses AraBERT (pretrained Arabic language model) to replace words with contextually appropriate alternatives, improving semantic diversity while preserving meaning.

## Advantages
- Preserves sentence context
- Generates diverse variations
- No manual dictionary needed

## Limitations
- Produces inconsistent results
- Often inserts illogical stopwords
- Requires GPU for reasonable speed
- Unreliable for Arabic text augmentation

```python
# ============================================================
# STEP 5.1: ARABERT CONTEXTUAL AUGMENTATION
# ============================================================

# Install required libraries
!pip install transformers torch -q

import pandas as pd
import numpy as np
import torch
import random
from transformers import AutoTokenizer, AutoModelForMaskedLM
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# CLASS: AraBERTAugmenter
# ============================================================

class AraBERTAugmenter:
    """
    Arabic contextual augmentation using AraBERT with [MASK] token prediction.
    Replaces words with contextually appropriate alternatives.
    """

    def __init__(self, model_name="aubmindlab/bert-base-arabertv2"):
        print(f"Loading model: {model_name}")
        print("This may take 2-3 minutes...")

        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModelForMaskedLM.from_pretrained(model_name)
        self.model.eval()

        # Move to GPU if available
        if torch.cuda.is_available():
            self.model = self.model.to('cuda')
            self.device = 'cuda'
        else:
            self.device = 'cpu'

        print(f"Model loaded successfully on {self.device.upper()}")

    def augment_sentence(self, text, aug_p=0.3):
        """
        Augment Arabic sentence using AraBERT with [MASK] token.

        Parameters:
        - text: Arabic sentence
        - aug_p: Percentage of words to mask (0-1)

        Returns:
        - Augmented sentence (or original if error occurs)
        """
        words = text.split()

        if len(words) < 4:
            return text

        # Select words to mask (avoid first and last word)
        num_to_mask = max(1, int(len(words) * aug_p))
        mask_indices = random.sample(range(1, len(words)-1), min(num_to_mask, len(words)-2))

        if not mask_indices:
            return text

        # Replace selected words with [MASK]
        masked_words = words.copy()
        for idx in mask_indices:
            masked_words[idx] = self.tokenizer.mask_token

        masked_text = ' '.join(masked_words)

        try:
            # Tokenize and predict
            inputs = self.tokenizer(masked_text, return_tensors='pt').to(self.device)

            with torch.no_grad():
                outputs = self.model(**inputs)
                logits = outputs.logits

            # Get predictions for each masked position
            mask_token_id = self.tokenizer.mask_token_id
            mask_positions = (inputs['input_ids'] == mask_token_id).nonzero(as_tuple=True)[1]

            new_words = words.copy()

            for idx, mask_pos in enumerate(mask_positions):
                # Get top predictions for this mask position
                logits_at_mask = logits[0, mask_pos, :]
                top_indices = torch.topk(logits_at_mask, 10).indices.tolist()

                original_word = words[mask_indices[idx]]

                # Find a suitable replacement
                for token_idx in top_indices:
                    predicted_token = self.tokenizer.decode([token_idx]).strip()

                    # Validate token
                    if (len(predicted_token) > 1 and
                        predicted_token not in ['[UNK]', '[CLS]', '[SEP]', '[PAD]', '[MASK]'] and
                        predicted_token.isalpha() and
                        predicted_token != original_word and
                        predicted_token not in ['و', 'ف', 'ثم', 'أو']):

                        new_words[mask_indices[idx]] = predicted_token
                        break

            return ' '.join(new_words)

        except Exception:
            return text

    def augment_batch(self, texts, aug_p=0.3):
        """
        Augment a batch of Arabic texts.

        Parameters:
        - texts: List of Arabic sentences
        - aug_p: Percentage of words to mask

        Returns:
        - List of augmented sentences
        """
        augmented = []

        for text in texts:
            try:
                aug_text = self.augment_sentence(text, aug_p)
                augmented.append(aug_text)
            except Exception:
                augmented.append(text)

        return augmented


# ============================================================
# TESTING
# ============================================================

print("=" * 60)
print("ARABERT CONTEXTUAL AUGMENTATION - TEST")
print("=" * 60)

# Initialize augmenter
augmenter = AraBERTAugmenter()
print("\nAugmenter ready for testing")

# Sample text
sample_text = "الطفل يقرأ كتاباً ممتعاً في المكتبة"
print(f"\nOriginal text:\n   {sample_text}")

# Generate augmented versions
print("\nGenerating augmented versions...")

augmented_texts = []
for i in range(3):
    augmented = augmenter.augment_sentence(sample_text, aug_p=0.3)
    augmented_texts.append(augmented)

print("\nAugmented texts:")
for i, text in enumerate(augmented_texts, 1):
    print(f"   {i}. {text}")

print("\n" + "=" * 60)
print("Test complete")
print("=" * 60)
```

# Step 5.2: Arabic Back Translation Augmentation

## Overview
Back translation generates paraphrased text by translating from Arabic to English and back to Arabic. This produces natural variations with different word choices and sentence structures while preserving semantic meaning.

**Process:**
- Arabic → English (pivot language)
- English → Arabic (back to target language)

## Advantages
- Preserves core meaning
- Produces grammatically correct Arabic
- Changes word order and synonyms naturally
- No manual dictionary required
- Widely used in NLP research

## Limitations
- Requires GPU for reasonable speed
- Slow for large datasets
- Quality depends on translation models
- Some cultural nuances may be lost

```python
# ============================================================
# STEP 5.2: ARABIC BACK TRANSLATION AUGMENTATION
# ============================================================

# Install required libraries
!pip install transformers torch sentencepiece -q

import pandas as pd
import numpy as np
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import warnings
warnings.filterwarnings('ignore')

# ============================================================
# LOAD TRANSLATION MODELS
# ============================================================

print("=" * 60)
print("BACK TRANSLATION AUGMENTATION - SETUP")
print("=" * 60)

# Model names
AR_TO_EN_MODEL = "Helsinki-NLP/opus-mt-ar-en"
EN_TO_AR_MODEL = "Helsinki-NLP/opus-mt-en-ar"

print(f"Loading Arabic to English model: {AR_TO_EN_MODEL}")
print(f"Loading English to Arabic model: {EN_TO_AR_MODEL}")
print("This may take 2-3 minutes...")

# Set device
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Load Arabic to English
ar_en_tokenizer = AutoTokenizer.from_pretrained(AR_TO_EN_MODEL)
ar_en_model = AutoModelForSeq2SeqLM.from_pretrained(AR_TO_EN_MODEL).to(device)

# Load English to Arabic
en_ar_tokenizer = AutoTokenizer.from_pretrained(EN_TO_AR_MODEL)
en_ar_model = AutoModelForSeq2SeqLM.from_pretrained(EN_TO_AR_MODEL).to(device)

print(f"Models loaded successfully on {device.upper()}")


# ============================================================
# BACK TRANSLATION FUNCTION
# ============================================================

def back_translate(text):
    """
    Perform back translation: Arabic -> English -> Arabic

    Parameters:
    - text: Arabic text to augment

    Returns:
    - Augmented Arabic text (or original if translation fails)
    """
    try:
        # Step 1: Arabic -> English
        inputs = ar_en_tokenizer(text, return_tensors="pt", truncation=True, max_length=512).to(device)
        outputs = ar_en_model.generate(
            **inputs,
            max_length=512,
            num_beams=4,
            early_stopping=True,
            no_repeat_ngram_size=3
        )
        english_text = ar_en_tokenizer.decode(outputs[0], skip_special_tokens=True)

        if not english_text.strip():
            return text

        # Step 2: English -> Arabic
        inputs = en_ar_tokenizer(english_text, return_tensors="pt", truncation=True, max_length=512).to(device)
        outputs = en_ar_model.generate(
            **inputs,
            max_length=512,
            num_beams=4,
            early_stopping=True,
            no_repeat_ngram_size=3
        )
        arabic_text = en_ar_tokenizer.decode(outputs[0], skip_special_tokens=True)

        return arabic_text if arabic_text.strip() else text

    except Exception:
        return text


# ============================================================
# TEST BACK TRANSLATION
# ============================================================

print("\n" + "=" * 60)
print("BACK TRANSLATION - TEST")
print("=" * 60)

# Sample Arabic texts
sample_texts = [
    "الطفل يقرأ كتاباً ممتعاً في المكتبة",
    "الطقس جميل اليوم وسنذهب إلى الحديقة",
    "التعلم الآلي يساعد في حل المشكلات المعقدة",
    "القراءة تنمي العقل وتوسع المدارك",
    "يجب علينا المحافظة على البيئة نظيفة"
]

print("\nOriginal texts:")
for i, text in enumerate(sample_texts, 1):
    print(f"   {i}. {text}")

# Apply back translation
print("\nPerforming back translation...")

augmented_texts = []
for i, text in enumerate(sample_texts):
    print(f"   Processing {i+1}/{len(sample_texts)}...")
    result = back_translate(text)
    augmented_texts.append(result)

# Display results
print("\n" + "-" * 60)
print("Results:")
print("-" * 60)

success_count = 0
for i, (original, augmented) in enumerate(zip(sample_texts, augmented_texts), 1):
    print(f"\n{i}. Original:")
    print(f"   {original}")
    print(f"   Augmented:")
    print(f"   {augmented}")

    if original != augmented:
        print(f"   Status: CHANGED")
        success_count += 1
    else:
        print(f"   Status: UNCHANGED")

# Summary
print("\n" + "=" * 60)
print("Test complete")
print(f"Successfully augmented: {success_count}/{len(sample_texts)}")
print(f"Success rate: {(success_count/len(sample_texts))*100:.0f}%")
print("=" * 60)
```
